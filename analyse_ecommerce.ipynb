```python
#Suppression des doublons
print(ecom_df.duplicated().sum())
```

```python
ecom_df = ecom_df.drop_duplicates()
```

```python
#Gestion des valeurs manquantes
print(ecom_df.isnull().sum())
```

```python
ecom_df.loc[ecom_df['Description'].isnull(), 'Description']='Inconnue'
ecom_df.loc[ecom_df['Country'].isnull(), 'Country']='Inconnue'
```

```python
ecom_df = ecom_df.dropna(subset=['StockCode'])
ecom_df = ecom_df.dropna(subset=['Quantity'])
ecom_df = ecom_df.dropna(subset=['UnitPrice'])
ecom_df = ecom_df.dropna(subset=['InvoiceDate'])
```
